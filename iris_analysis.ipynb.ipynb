```python
import pandas as pd

# 1. Read the CSV file into a DataFrame
df = pd.read_csv('iris-analysis/data/iris.csv')

# 2. Display the first 5 rows
print("--- First 5 Rows ---")
display(df.head())

# 3. Display the last 5 rows
print("\n--- Last 5 Rows ---")
display(df.tail())

# 4. Display rows and columns count
print(f"\nShape of DataFrame (Rows, Columns): {df.shape}")

# 5. Display column names
print(f"\nColumn Names: {list(df.columns)}")

# 6. Display data types
print("\nData Types:")
print(df.dtypes)
```

```python
# Task 4: Explore and Analyze the Dataset

# 1 & 2. Missing values check
print("--- Missing Values in Each Column ---")
print(df.isnull().sum())

# 3. Duplicate rows check
print(f"\nNumber of duplicate rows: {df.duplicated().sum()}")

# 4. Unique species count
print(f"\nNumber of unique species: {df['species'].nunique()}")

# 5. Flowers count per species
print("\n--- Flower Count per Species ---")
print(df['species'].value_counts())

# 6, 7, 8. Specific metrics
print(f"\nAverage Sepal Length: {df['sepal_length'].mean():.2f}")
print(f"Maximum Petal Length: {df['petal_length'].max():.2f}")
print(f"Minimum Sepal Width: {df['sepal_width'].min():.2f}")

# 9. Groupby species mean
print("\n--- Mean of Numeric Columns per Species ---")
species_summary = df.groupby('species').mean()
display(species_summary)

# 10. Descriptive statistics
print("\n--- Descriptive Statistics ---")
display(df.describe())
```

```python
# Task 5: Filter and Analyze Data

# 1. Display flowers where sepal_length > 6
print("--- Flowers with Sepal Length > 6 ---")
filtered_sepal = df[df['sepal_length'] > 6]
display(filtered_sepal.head())

# 2. Display flowers where petal_width > 1.5
print("\n--- Flowers with Petal Width > 1.5 ---")
filtered_petal = df[df['petal_width'] > 1.5]
display(filtered_petal.head())

# 3. Display setosa species only
print("\n--- Setosa Species Only ---")
setosa_df = df[df['species'] == 'setosa']
display(setosa_df.head())

# 4. Species with the highest average petal length
highest_petal_species = df.groupby('species')['petal_length'].mean().idxmax()
print(f"\nSpecies with highest average petal length: {highest_petal_species}")

# 5. Sort dataset by sepal_length in descending order
print("\n--- Dataset Sorted by Sepal Length (Descending) ---")
sorted_df = df.sort_values(by='sepal_length', ascending=False)
display(sorted_df.head())
```

```python
# Task 6: Export Analysis Results

import os

# 1. Create the output folder if it doesn't exist
os.makedirs('iris-analysis/output', exist_ok=True)

# 2. Save species-level summary statistics
species_summary.to_csv('iris-analysis/output/species_summary.csv')

# 3. Save filtered flowers where sepal_length > 6
filtered_sepal.to_csv('iris-analysis/output/filtered_iris.csv', index=False)

# 4. Confirm creation
print("Files exported successfully!")
print("1. output/species_summary.csv")
print("2. output/filtered_iris.csv")
```
